```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        a = {}
        start = 0
        max_len = 0
        for i , c in enumerate(s):
            if c in a and a[c]>= start:
                start = a[c] + 1 
            a[c]=i 
            max_len= max(max_len, i - start + 1) 
        return max_len
        
```
